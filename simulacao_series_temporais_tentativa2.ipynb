## Suavização exponencial sem tendência e sem sazonalidade.



```python
# Instalação opcional no Google Colab (descomente se necessário)
# !pip install statsmodels plotly pandas numpy

# Importa o NumPy para gerar números aleatórios e fazer operações numéricas.
import numpy as np
# Importa o pandas para organizar a série temporal em DataFrame.
import pandas as pd
# Importa o Plotly para criar gráficos interativos.
import plotly.graph_objects as go
# Importa o modelo de suavização exponencial do statsmodels.
from statsmodels.tsa.holtwinters import ExponentialSmoothing

# Define uma semente para que a simulação possa ser reproduzida.
np.random.seed(123)
# Define o número de observações simuladas.
n = 120
# Cria um índice de tempo mensal para a série.
datas = pd.date_range(start='2015-01-01', periods=n, freq='M')
# Define o nível verdadeiro da série simulada.
nivel = 50
# Gera ruídos aleatórios com média zero e desvio padrão 3.
erros = np.random.normal(loc=0, scale=3, size=n)
# Constrói a série como um nível constante mais ruído.
serie = nivel + erros
# Organiza os dados em um DataFrame com datas e valores.
df = pd.DataFrame({'data': datas, 'y': serie})
# Define a coluna de datas como índice do DataFrame.
df = df.set_index('data')

# Cria o modelo de suavização exponencial simples, sem tendência e sem sazonalidade.
modelo = ExponentialSmoothing(df['y'], trend=None, seasonal=None, initialization_method='estimated')
# Ajusta o modelo aos dados simulados.
resultado = modelo.fit()
# Gera previsões para os próximos 10 períodos.
previsao = resultado.forecast(10)

# Cria uma tabela com os parâmetros estimados do modelo.
parametros = pd.DataFrame({
    'parametro': list(resultado.params.keys()),
    'estimativa': list(resultado.params.values())
})
# Mostra a tabela de parâmetros no console.
print(parametros)

# Cria o primeiro gráfico com a série original.
fig1 = go.Figure()
# Adiciona a linha da série observada ao primeiro gráfico.
fig1.add_trace(go.Scatter(x=df.index, y=df['y'], mode='lines', name='Série original'))
# Define o layout do primeiro gráfico.
fig1.update_layout(title='Suavização exponencial simples: série original', xaxis_title='Tempo', yaxis_title='Valor', template='plotly_white')
# Exibe o primeiro gráfico.
fig1.show()

# Cria o segundo gráfico com série original e previsão.
fig2 = go.Figure()
# Adiciona a série observada ao segundo gráfico.
fig2.add_trace(go.Scatter(x=df.index, y=df['y'], mode='lines', name='Série original'))
# Adiciona os 10 valores previstos ao segundo gráfico.
fig2.add_trace(go.Scatter(x=previsao.index, y=previsao.values, mode='lines+markers', name='Previsão 10 passos'))
# Define o layout do segundo gráfico.
fig2.update_layout(title='Suavização exponencial simples: série e previsão', xaxis_title='Tempo', yaxis_title='Valor', template='plotly_white')
# Exibe o segundo gráfico.
fig2.show()
```

## Suavização exponencial com tendência e sem sazonalidade.


```python
# Instalação opcional no Google Colab (descomente se necessário)
# !pip install statsmodels plotly pandas numpy

# Importa o NumPy para geração de tendência com ruído.
import numpy as np
# Importa o pandas para estruturar a série temporal.
import pandas as pd
# Importa o Plotly para visualização gráfica.
import plotly.graph_objects as go
# Importa o modelo de suavização exponencial.
from statsmodels.tsa.holtwinters import ExponentialSmoothing

# Fixa a semente para reprodutibilidade.
np.random.seed(456)
# Define o tamanho da série.
n = 120
# Cria datas mensais.
datas = pd.date_range(start='2015-01-01', periods=n, freq='M')
# Cria um vetor de tempo inteiro.
t = np.arange(n)
# Define nível inicial da série.
intercepto = 20
# Define a inclinação da tendência linear.
beta = 0.8
# Gera ruído aleatório.
erros = np.random.normal(loc=0, scale=4, size=n)
# Constrói a série com tendência linear e ruído.
serie = intercepto + beta * t + erros
# Cria o DataFrame com os dados simulados.
df = pd.DataFrame({'data': datas, 'y': serie}).set_index('data')

# Cria o modelo de Holt com tendência aditiva e sem sazonalidade.
modelo = ExponentialSmoothing(df['y'], trend='add', seasonal=None, initialization_method='estimated')
# Ajusta o modelo aos dados.
resultado = modelo.fit()
# Faz previsão de 10 períodos futuros.
previsao = resultado.forecast(10)

# Monta a tabela de parâmetros estimados.
parametros = pd.DataFrame({
    'parametro': list(resultado.params.keys()),
    'estimativa': list(resultado.params.values())
})
# Imprime a tabela no console.
print(parametros)

# Cria o gráfico da série original.
fig1 = go.Figure()
# Insere a linha da série simulada.
fig1.add_trace(go.Scatter(x=df.index, y=df['y'], mode='lines', name='Série original'))
# Ajusta título e eixos do gráfico original.
fig1.update_layout(title='Holt com tendência: série original', xaxis_title='Tempo', yaxis_title='Valor', template='plotly_white')
# Mostra o gráfico original.
fig1.show()

# Cria o gráfico com série e previsões.
fig2 = go.Figure()
# Adiciona a série observada.
fig2.add_trace(go.Scatter(x=df.index, y=df['y'], mode='lines', name='Série original'))
# Adiciona a previsão de 10 passos.
fig2.add_trace(go.Scatter(x=previsao.index, y=previsao.values, mode='lines+markers', name='Previsão 10 passos'))
# Ajusta o layout do gráfico final.
fig2.update_layout(title='Holt com tendência: série e previsão', xaxis_title='Tempo', yaxis_title='Valor', template='plotly_white')
# Exibe o gráfico com previsão.
fig2.show()
```

## Suavização exponencial com tendência e sazonalidade.



```python
# Instalação opcional no Google Colab (descomente se necessário)
# !pip install statsmodels plotly pandas numpy

# Importa o NumPy para gerar componentes determinísticos e ruído.
import numpy as np
# Importa o pandas para manipular dados temporais.
import pandas as pd
# Importa o Plotly para os gráficos interativos.
import plotly.graph_objects as go
# Importa o modelo de suavização exponencial com sazonalidade.
from statsmodels.tsa.holtwinters import ExponentialSmoothing

# Define a semente do gerador aleatório.
np.random.seed(789)
# Define o número de observações da série.
n = 144
# Cria datas mensais para 12 anos.
datas = pd.date_range(start='2010-01-01', periods=n, freq='M')
# Cria o índice do tempo.
t = np.arange(n)
# Define o nível inicial.
intercepto = 30
# Define a tendência linear.
beta = 0.5
# Define o período sazonal mensal.
periodo = 12
# Cria uma componente sazonal senoidal.
sazonal = 8 * np.sin(2 * np.pi * t / periodo)
# Gera ruído aleatório normal.
erros = np.random.normal(loc=0, scale=2.5, size=n)
# Soma nível, tendência, sazonalidade e ruído.
serie = intercepto + beta * t + sazonal + erros
# Organiza a série em DataFrame indexado por data.
df = pd.DataFrame({'data': datas, 'y': serie}).set_index('data')

# Define o modelo Holt-Winters com tendência aditiva e sazonalidade aditiva.
modelo = ExponentialSmoothing(df['y'], trend='add', seasonal='add', seasonal_periods=12, initialization_method='estimated')
# Ajusta o modelo aos dados simulados.
resultado = modelo.fit()
# Gera previsão para 10 períodos à frente.
previsao = resultado.forecast(10)

# Cria a tabela de parâmetros estimados.
parametros = pd.DataFrame({
    'parametro': list(resultado.params.keys()),
    'estimativa': list(resultado.params.values())
})
# Imprime a tabela de parâmetros.
print(parametros)

# Cria o gráfico da série original.
fig1 = go.Figure()
# Adiciona a série ao gráfico.
fig1.add_trace(go.Scatter(x=df.index, y=df['y'], mode='lines', name='Série original'))
# Personaliza o layout do gráfico original.
fig1.update_layout(title='Holt-Winters com tendência e sazonalidade: série original', xaxis_title='Tempo', yaxis_title='Valor', template='plotly_white')
# Mostra o gráfico da série original.
fig1.show()

# Cria o gráfico combinando série observada e previsão.
fig2 = go.Figure()
# Adiciona a parte observada da série.
fig2.add_trace(go.Scatter(x=df.index, y=df['y'], mode='lines', name='Série original'))
# Adiciona os valores previstos.
fig2.add_trace(go.Scatter(x=previsao.index, y=previsao.values, mode='lines+markers', name='Previsão 10 passos'))
# Ajusta o layout do gráfico com previsão.
fig2.update_layout(title='Holt-Winters com tendência e sazonalidade: série e previsão', xaxis_title='Tempo', yaxis_title='Valor', template='plotly_white')
# Exibe o gráfico final.
fig2.show()
```

## ARIMA(2,0,0)



```python
# Instalação opcional no Google Colab (descomente se necessário)
# !pip install statsmodels plotly pandas numpy
# Importa o NumPy para gerar a série simulada.
import numpy as np
# Importa o pandas para armazenar a série com datas.
import pandas as pd
# Importa o Plotly para visualização.
import plotly.graph_objects as go
# Importa o modelo ARIMA do statsmodels.
from statsmodels.tsa.arima.model import ARIMA

# Define a semente aleatória para reproduzir os resultados.
np.random.seed(321)
# Define o número de observações da série simulada.
n = 150
# Cria um índice mensal para a série temporal.
datas = pd.date_range(start='2012-01-01', periods=n, freq='M')
# Define a estrutura do modelo ARIMA que será usada na simulação.
modelo_simulacao = ARIMA(endog=np.zeros(n), order=(2, 0, 0))
# Define um vetor de parâmetros estáveis para a simulação do modelo.
params = np.array([0.6, -0.25, 1.0, 1.0]) # Adicionado o parâmetro de variância (sigma2)
# Gera uma nova série temporal simulada a partir do modelo especificado.
serie = modelo_simulacao.simulate(params=params, nsimulations=n)
# Converte a série simulada em DataFrame indexado por data.
df = pd.DataFrame({'data': datas, 'y': np.asarray(serie)}).set_index('data')

# Cria o modelo ARIMA para ajustar à série simulada.
modelo = ARIMA(df['y'], order=(2, 0, 0))
# Ajusta o modelo ARIMA aos dados observados.
resultado = modelo.fit()
# Produz previsão para os próximos 10 períodos.
previsao = resultado.forecast(10)

# Extrai os parâmetros estimados do modelo ajustado.
parametros = pd.DataFrame({
    'parametro': resultado.param_names,
    'estimativa': resultado.params
})
# Imprime a tabela de parâmetros no console.
print(parametros)

# Cria o gráfico da série original.
fig1 = go.Figure()
# Adiciona a linha da série observada.
fig1.add_trace(go.Scatter(x=df.index, y=df['y'], mode='lines', name='Série original'))
# Ajusta o layout do gráfico da série original.
fig1.update_layout(title='ARIMA(2,0,0): série original', xaxis_title='Tempo', yaxis_title='Valor', template='plotly_white')
# Exibe o gráfico original.
fig1.show()

# Cria o gráfico com série e previsão futura.
fig2 = go.Figure()
# Adiciona a série observada ao gráfico final.
fig2.add_trace(go.Scatter(x=df.index, y=df['y'], mode='lines', name='Série original'))
# Adiciona os 10 valores previstos ao gráfico final.
fig2.add_trace(go.Scatter(x=previsao.index, y=previsao.values, mode='lines+markers', name='Previsão 10 passos'))
# Personaliza o gráfico de série com previsão.
fig2.update_layout(title='ARIMA(2,0,0): série e previsão', xaxis_title='Tempo', yaxis_title='Valor', template='plotly_white')
# Exibe o gráfico final.
fig2.show()
```
